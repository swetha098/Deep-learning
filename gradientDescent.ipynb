1.2.1 Loss functions and Gradients


```python
#loss function implementation

import numpy as np

def Loss_fn(x,y,w,b):
  M=len(x)
  y_pred=w*x+b
  error=y_pred-y
  Loss=np.sum(error**2)/(2*M)

  return Loss
```

```python
#Test code

x=np.array([1,2,3])
y=np.array([2,4,6])
w=0.5
b=0.5
Loss_fn(x,y,w,b)

print("Loss:",Loss_fn(x,y,w,b))
```

b) gradient w.r.t w


$$y=wx+b$$
substituting in the given loss function:
$$
L(w,b) = \frac{1}{2M}\sum_{i=1}^{M}(wx_i+b-y_i)^2
$$

$$
\frac{\partial L}{\partial w}
=
\frac{\partial}{\partial w}
\left[
\frac{1}{2M}\sum_{i=1}^{M}(wx_i+b-y_i)^2
\right]
$$

$$
\frac{\partial L}{\partial w}
=
\frac{1}{2M}
\sum_{i=1}^{M}
2(wx_i+b-y_i)x_i
$$
$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{1}{M}
\sum_{i=1}^{M}(wx_i+b-y_i)x_i
}
$$
$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{1}{M}
\sum_{i=1}^{M}(\hat{y}_i-y_i)x_i
}
$$



```python
def dw(x,y,w,b):
  M=len(x)
  y_pred=w*x+b
  error=y_pred-y
  dw=np.sum(error*x)/M
  return dw
```

```python
#test code

gradient=dw(x,y,w,b)
print("Gradient:",gradient)
```
